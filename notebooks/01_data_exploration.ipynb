## Missing values / infinities check

This is what `preprocessing.clean_data()` handles automatically in the pipeline -- worth checking manually here first to understand the raw data quality.

```python
import numpy as np

nan_counts = df[config.FEATURE_COLUMNS].isna().sum()
inf_counts = np.isinf(df[config.FEATURE_COLUMNS].select_dtypes(include=[np.number])).sum()

print("Columns with NaN values:\n", nan_counts[nan_counts > 0])
print("\nColumns with inf values:\n", inf_counts[inf_counts > 0])
```
