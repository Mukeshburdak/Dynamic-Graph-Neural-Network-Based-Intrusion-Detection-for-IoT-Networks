## Train a single model (quick demo: few epochs)

For a real run, remove `num_epochs` to use the full `config.NUM_EPOCHS` with early stopping.

```python
history_gcn = train_model("gcn", num_epochs=10)
history_gcn.tail()
```

```python
fig = plot_training_curves(history_gcn, model_name="gcn")
```

## Train the remaining models

```python
history_sage = train_model("graphsage", num_epochs=10)
fig = plot_training_curves(history_sage, model_name="graphsage")
```

```python
history_gat = train_model("gat", num_epochs=10)
fig = plot_training_curves(history_gat, model_name="gat")
```

## Compare validation macro-F1 across models so far

```python
import pandas as pd

summary = pd.DataFrame({
    "model": ["GCN", "GraphSAGE", "GAT"],
    "best_val_macro_f1": [
        history_gcn["val_macro_f1"].max(),
        history_sage["val_macro_f1"].max(),
        history_gat["val_macro_f1"].max(),
    ],
})
summary.sort_values("best_val_macro_f1", ascending=False)
```
